What is a String?

A string is a sequence of characters enclosed in:

" " double quotes
' ' single quotes

```python
s ='srushti'
print(s)
```

```python
print(s[0])
print(s[-1])
print(s[-2])
```

String Slicing

Syntax:s[start : end : step]

```python
print(s[0:4])
print(s[::-1]) 
print(s[2::])
```

string are immutable you cannot change the character directly

```python
s="srushti"
s="S"+s[1:]
print(s)
```
