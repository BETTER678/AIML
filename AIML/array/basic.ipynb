```python
#sum of array
arr=[1,2,3,3,4,5,6]
total = 0
for num in arr:
    total += num
print(total)
```

```python
nums = [2, 7, 11, 15]
target = 9

d = {}
for i in range(len(nums)):
    if target - nums[i] in d:
        print(d[target - nums[i]], i)
    d[nums[i]] = i
```

```python
#Count even numbers
arr= [2,3,4,5,55,66,51]
count=0
for i in arr:
    if i%2==0:
        count=count+1
print(count)
 
```

```python
#remove the dublicates from the array
list=[1,2,3,3,4,4,4,5,5,6,7,7]
unique=[]
for i in list:
    if i not in unique:
        unique.append(i)
print(unique)
```
